# Proyecto Mineria de datos

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

pd.options.d["display"]["max_columns"] = None
```

```python
df = pd.read_csv("datos/datos.csv")
df.head()
```

```python
# Graficamos la distribución de english 
plt.hist(df["english"].dropna(), bins=2, alpha=0.7)
plt.title("Distribución de los juegos en ingles")

plt.xlabel("Inglés (f = No, t = Sí)")
plt.ylabel("Número de juegos")

# Calculamos el porcentaje de juegos en inglés
total_juegos = len(df)
juegos_ingles = df["english"].value_counts().get(True, 1)
print(juegos_ingles / total_juegos * 100)
```

```python
# Graficamos según rating

plt.figure()
sns.histplot(df["reviewscore"].dropna(), bins=20, alpha=0.7, palette="mako")
plt.title("Distribución del puntaje de reseñas")

# sns.countplot(x='reviewscore', data=df, palette="mako")
# sns.barplot(x=, y="reviewscore", data=df, palette="mako")
plt.show()
```

```python
# Graficamos según rating

plt.figure()
sns.histplot(df["copiessold"].dropna(), bins=20, alpha=0.7, palette="mako")
plt.title("Distribución del puntaje de reseñas")

plt.show()
```

```python


#print(df["release_date"].head())

df["mes"] = pd.to_datetime(df["release_date"]).dt.month
df["mes"].head()
```

```python
# Age
var = 'reviewscore'
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x=var, hue='mes', multiple='stack', bins=30, palette='mako')
plt.title(f'Distribución de {var} segmentada por mes de lanzamiento')
plt.xlabel(var)
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

```python
# Primero filtraremos en los datos que tengan un revenue mayor a 1000

"""filtrado = df['revenue']
df_filtrado = df[filtrado]"""

# Age

xvar=np.log10(df['revenue']+1)


var = 'revenue'
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x=xvar, hue='mes', multiple='stack', bins=30, palette='mako')
# plt.yscale('log')
plt.title(f'Distribución de {var} segmentada por mes de lanzamiento')
plt.xlabel(f"log_10({var}+1)")
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

```python
# Primero filtraremos en los datos que tengan un revenue mayor a 1000

var = 'copiessold'
xvar=np.log10(df[var]+1)


plt.figure(figsize=(8, 4))
sns.histplot(data=df, x=xvar, hue='mes', multiple='stack', bins=30, palette='mako')
# plt.yscale('log')
plt.title(f'Distribución de {var} segmentada por mes de lanzamiento')
plt.xlabel(f"log_10({var}+1)")
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```

```python
var = 'median_playtime'
xvar= np.log10(np.log10(df[var]+1))


plt.figure(figsize=(8, 4))
sns.histplot(data=df, x=xvar, hue='mes', multiple='stack', bins=30, palette='mako')
# plt.yscale('log')
plt.title(f'Distribución de {var} segmentada por mes de lanzamiento')
plt.xlabel(f"log_10(log_10({var}+1))")
plt.ylabel('Frecuencia')
plt.tight_layout()
plt.show()
```
